```python
#####4- Prime Star Systems

arr = np.array([
    [2,3,5],
    [14,6,8],
    [11, 13, 17],
    [7,10,13]
])

def checkprime(num): 
    if num < 2:
        return False
    for i in range(2, int(num**0.5)+1):
        if num % i == 0:
            return False
    return True

def primenum_rows(arr):
    new_arr = []
    for row in arr:
        if any(checkprime(num) for num in row):
                new_arr.append(row)
    return np.array(new_arr)

print(primenum_rows(arr))
```

```python
#####5- Let's Play Checkers
def original_checkerboard(x,y): 
    return np.zeros(shape = (x,y))

print(original_checkerboard(8,8))
```

```python
#combined 5.2 and 5.3 here! hope that's ok

def checkerboard(): 
    checkerboard = original_checkerboard(8,8)
    for row in range(0,8):
        if row % 2 == 0:
            for i in range(0,7,2):
                checkerboard[row][i] = 1
        elif row % 2 == 1:
            for i in range(1,7,2):
                checkerboard[row][i] = 1
    return checkerboard
                
checkerboard()
```

```python
def reverse_checkerboard():
    checkerboard = original_checkerboard(8,8)
    for row in range(0,8):
        if row % 2 == 1:
            for i in range(0,7,2):
                checkerboard[row][i] = 1
        elif row % 2 == 0:
            for i in range(1,7,2):
                checkerboard[row][i] = 1
    return checkerboard

reverse_checkerboard()
```

```python
######6- Expanding Universe
```

```python
#inputting an array of strings and an intenger, return the strings with intenger-specified number of spaces between the letters
def expansion(arr, intenger):
    number_of_spaces= []
    empty_list= []
    for i in range(1,intenger):
        number_of_spaces.append(' ')
    number_of_spaces_but_a_string = ''.join(number_of_spaces)
    for string in arr:
        char_list = list(string)
        new_list = []
        for letter in char_list[:-1]:
            new_list.append(letter + number_of_spaces_but_a_string)
        new_list.append(char_list[-1])
        final_string= ''.join(new_list)
        empty_list.append(final_string) 
    final_array = np.array(empty_list)
    print(final_array)
            
arr= np.array(["galaxy", "clusters"])
expansion(arr,5)
```

```python
def secondBrightest(array):
    empty_list = []
    for row in array:
        brightest = row[0]
        second_brightest = row[0]
        for number in row:
            if number > brightest:
                brightest = number
        for number in row:
            if number < brightest and number> second_brightest:
                second_brightest = number
        empty_list.append(second_brightest)
    final_array = np.array(empty_list)
    print(final_array)
                
np.random.seed(123)
stars = np.random.randint(500, 2000, (5, 5))
print(stars)


secondBrightest(stars)
```
