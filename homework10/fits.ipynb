```python
#2
import matplotlib.pyplot as plt
from astropy.io import fits
import numpy as np
```

```python
#3
header = []
data = []

for i in range(1,6):
    with fits.open(f"../../course_assignments/homework10/M13_000{i}.fits") as hdul:
        header.append(hdul[0].header)
        data.append(hdul[0].data)
```

```python
#PLOT IMAGES
```

```python
#1
plt.figure(figsize=(20,10))
fig, axes = plt.subplots(1,5, figsize = (20,5))

axes[0].imshow(data[0], origin="lower")
axes[1].imshow(data[1], origin= "lower")
axes[2].imshow(data[2], origin="lower")
axes[3].imshow(data[3], origin="lower")
axes[4].imshow(data[4], origin="lower")

plt.show()
```

```python
#2
plt.figure(figsize=(20,10))
fig, axes = plt.subplots(1,5, figsize = (20,5))

vmin = 500
vmax = 1000

axes[0] = axes[0].imshow(data[0], origin="lower", vmin=vmin, vmax=vmax)
axes[1] = axes[1].imshow(data[1], origin= "lower", vmin=vmin, vmax=vmax)
axes[2] = axes[2].imshow(data[2], origin="lower", vmin=vmin, vmax=vmax)
axes[3] = axes[3].imshow(data[3], origin="lower", vmin=vmin, vmax=vmax)
axes[4]= axes[4].imshow(data[4], origin="lower", vmin=vmin, vmax=vmax)

for i in range(5):
    plt.colorbar(axes[i], shrink= 0.5)

#These images all have the same color scheme, and they produce a relatively cleaner image of the object than the previous plots. 

plt.show()
```

```python
#PLOT HISTOGRAMS#
```

```python
#1
for file in data:
    plt.figure(figsize =(3,3))
    plt.xlabel("Value")
    plt.ylabel("Frequency")
    plt.title("histogram values")
    plt.hist(file.ravel(), bins=50)
```

```python
#2
#the x axis shows the pixel values/intensity 
#the y axis shows the pixel count

#bins=50 means that there are 50 bars of data; the data is sorted into 50 groups

print(np.median(data))
#Most of the data is centered at around 500 for pixel intensity (total median is 511)

#The data is not very spread out, as most of it is contained within a range of about 200, between 400 and 600. 

for file in data:
    print(np.median(file), np.mean(file))

#image 1 median: 512, mean:510.47
#image 2: median: 511, mean: 509.746
#image 3: median: 510, mean:508.050
#image 4: median: 509, mean: 507.906
#image 5: median: 510, mean: 508.792
```

```python
#COMBINE AND NORMALIZE IMAGES
```

```python
#1
combined_data = np.median(data, axis=0)

print(combined_data)
print(np.median(combined_data))

#the median intensity value of my combined image is 511. 
```

```python
#2
plt.hist(combined_data.ravel(), bins=50)
plt.title("Pixel Values")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

```python
#3
vmin = 470
vmax = 560

plt.figure(figsize=(10,10))
plt.imshow(combined_data, origin= "lower", vmin=vmin, vmax= vmax)
plt.colorbar(shrink=0.5)
plt.show()
```

```python
#4
#This image is much more clear than earlier versions! The star is visible at the center. 
```
