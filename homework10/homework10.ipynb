```python
#APPLY ASTROPY UNITS#
```

```python
#1
def attach_units(column_name, unit, new_col = None): #this makes new column name an optional input
    new_values = []
    for val in df[column_name]:
        new_values.append(val*unit)

    df[column_name] = new_values
    if new_col != None:
        df.rename(columns={column_name: new_col}, inplace=True)
```

```python
#2
attach_units("Orbital Period (days)", u.day, None)

print(df["Orbital Period (days)"])
```

```python
#3
#AU- astronomical unit, the avergage distance between the Earth and the Sun, a unit of measurement
print(const.au)
```

```python
print(const.au.to(u.km))
#There are 149597870.7 km in an AU
```

```python
def converting_to_AU(column_name, unit):
    new_values = []
    for value in df[column_name]:
        value_converted= value.to(u.au)
        new_values.append(value_converted)
        
    df[column_name] = new_values
```

```python
attach_units('Diameter (km)', u.km)
converting_to_AU('Diameter (km)', u.km)
```

```python
attach_units('Distance from Sun (10^6 km)', 1e6*u.km, 'Distance from Sun (km)' )
converting_to_AU('Distance from Sun (km)', u.km)

converting_to_AU('Perihelion (km)', u.km)

converting_to_AU('Aphelion (km)', u.km)

converting_to_AU('Semi_major_axis (km)', u.km)
```
