```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#1
def make_sine_wave(x,A,w):
    results= []
    for i in x:
        y = A*np.sin(w*i)
        results.append(y)
    return results
```

```python
#2
x= np.linspace(0,2*np.pi, 1000)

#3
amplitude = np.array([0.5, 1, 1.5, 2, 2.5])

frequency = np.array([1,2,3,4,5])

#4
fig,axes = plt.subplots(1,3, figsize = (25,5))

#5
for amp, freq in zip(amplitude, frequency):
    print(amp, freq)

#6
axes[0].plot(x, make_sine_wave(x, amplitude, 1), label = amplitude)
axes[0].set_title( "Sine of x, varying amplitude")
axes[0].set_xlabel("x-axis")
axes[0].set_ylabel("y-axis")
axes[0].legend()

axes[1].plot(x, make_sine_wave(x, 1, frequency), label = frequency)
axes[1].set_title( "Sine of x, varying frequency")
axes[1].set_xlabel("x-axis")
axes[1].set_ylabel("y-axis")
axes[1].legend()


for amp, freq in zip(amplitude, frequency):
    y = make_sine_wave(x, amp, freq)
    axes[2].plot(x, y, label = f"{amp},{freq}")
axes[2].set_title( "Sine of x, varying amplitude and frequency")
axes[2].set_xlabel("x-axis")
axes[2].set_ylabel("y-axis")
axes[2].legend()

plt.rcParams['font.size'] = 20


plt.show()
```
