```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt("/Users/jeevikarajasekar/python_decal_fa25/course_assignments/homework7/star_data.txt", skiprows=1)
data
```

```python
# ____Analyzing Imported Data_____ #
```

```python
np.shape(data)
#There are 2642 rows of data
```

```python
magnitude = data[:,0]
temperature = data[:,1]
```

```python
print(np.min(magnitude))
print(np.argmin(magnitude))
#min occurs at Star 1988

print(np.max(magnitude))
print(np.argmax(magnitude))
#max occurs  at Star 264
```

```python
print(np.min(temperature))
print(np.argmin(temperature))
#min occurs at Star 1923

print(np.max(temperature))
print(np.argmax(temperature))
#max occurs  at Star 2613
```

```python
#Star 1988 is the brightest (smallest magnitude), Star 264 is the dimmest (largest magnitude)
```

```python
# ____Plot and HR Diagram_____ #
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)
```

```python
plt.figure(figsize=(8,6))
plt.scatter(log_temperature, log_magnitude, label = "stars" )
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

#brightest
plt.plot(log_temperature[1988], log_magnitude[1988], marker = '*', color = "black", markersize = 10, label = "Brightest star")

#dimmest
plt.plot(log_temperature[264], log_magnitude[264], marker = '*', color = "purple", markersize = 10, label = "Dimmest Star")

#hottest
plt.plot(log_temperature[2613], log_magnitude[2613], marker = '*', color = "green", markersize = 10, label = "Hottest Star")

#coldest
plt.plot(log_temperature[1923], log_magnitude[1923], marker = '*', color = "orange", markersize = 10, label = "Coldest Star")

plt.legend(loc = "lower left", fontsize = 10)

plt.rcParams['font.size'] = 14 
plt.title("Hertzsprung-Russel Diagram") 
plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude")

plt.show()
```
