```python
#6
median_latitude = df["latitude"].median()
median_longitude = df["longitude"].median()

print(median_latitude)
print(median_longitude)

#median latitude is 39.4355, median longitude is -89.0932
```

```python
#7
absolute_distance_medianlatitude = []
for latitude in df["latitude"]:
    absolute_distance_medianlatitude.append(abs(latitude-median_latitude))

df["Absolute Distance to Median Latitude"]= absolute_distance_medianlatitude
```

```python
#8
absolute_distance_medianlongitude = []
for longitude in df["longitude"]:
    absolute_distance_medianlongitude.append(abs(longitude-median_longitude))

df["Absolute Distance to Median Longitude"]= absolute_distance_medianlongitude
```

```python
#9
df.shape
#I have 52 rows and 6 columns in the dataframe now. 
```

```python
#10
print(f"The state closest to the median latitude is {df.loc[df["Absolute Distance to Median Latitude"].idxmin(),"name"]} by {df["Absolute Distance to Median Latitude"].min():.2f} deg")
print(f"The state farthest to the median latitude is {df.loc[df["Absolute Distance to Median Latitude"].idxmax(),"name"]} by {df["Absolute Distance to Median Latitude"].max():.2f} deg")

print(f"The state closest to the median longitude is {df.loc[df["Absolute Distance to Median Longitude"].idxmin(),"name"]} by {df["Absolute Distance to Median Longitude"].min():.2f} deg ")
print(f"The state farthest to the median longitude is {df.loc[df["Absolute Distance to Median Longitude"].idxmax(),"name"]} by {df["Absolute Distance to Median Longitude"].max():.2f} deg ")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#4
df1 = pd.DataFrame(sns.load_dataset("dowjones"))
```

```python
#5
df1.dropna(axis=1)
```

```python
#6
print(df1.shape)
#There are 649 rows and 2 columns in the dataset

print(df1.head())
print(df1.tail())
#Columns:
# Date- Represents the date that the data was taken
# Price- Represents the price of the Dow Jones index at the specified date
```

```python
#7
fig,axes= plt.subplots(2,2)

axes[0][0].scatter(df1["Date"], df1["Price"])
axes[0][1].scatter(df1["Price"], df1["Date"])
sns.histplot(df1["Price"], ax= axes[1][0])
sns.boxplot(df1["Date"], ax = axes[1][1])

plt.tight_layout()
plt.show()
```

```python
fig,axes= plt.subplots(2,2)


axes[0][0].scatter(df1["Date"], df1["Price"], label="Price", color = "pink")
axes[0][0].set_title( "Dow Jones Price over Time")
axes[0][0].set_xlabel("date (by Year)")
axes[0][0].set_ylabel("Price (USD)")
axes[0][0].legend()

axes[0][1].scatter(df1["Price"], df1["Date"], color = "Chartreuse")
axes[0][1].set_title( "Time passsage over changing Dow Jones Price" )
axes[0][1].set_xlabel("Price (USD")
axes[0][1].set_ylabel("date (by Year)")

ax1 = sns.histplot(df1["Price"], ax= axes[1][0], color = "cyan")
ax1.set_title("Histogram of Dow Jones Prices")
ax1.set_xlabel("Price (USD)")
ax1.set_ylabel("Count")

ax2 = sns.boxplot(df1["Date"], ax = axes[1][1], color = "lavender")
ax2.set_title("Boxplot of Dates from 1914 to 1968")
ax2.set_xlabel("graph of dates")
ax2.set_ylabel("date (by year)")

plt.tight_layout()
plt.show()
```

```python
#9
#The first subplot makes the most sense. Time is an independent variable, and Price is shown to change dependent on time. 
#I learned that prices have been increasing since around 1914, except for a dip during what I think was the Great Depression. Prices began increasing at a much higher rate around 1950.

#The second subplot does not make much sense, but it was my only choice. Here, time is the dependent variable, while Price is the independent variable, which makes no sense.
#I learned that date increases at a constant rate, regardlesss of changing Dow Jones prices. The world keeps spinning and orbiting the sun...

#The third subplot shows the counts of various Dow Jones Price Ranges. It made more sense to make a histogram of Price rather than date, because a histogram of dates would show the same count for every date range!
#I learned that the prices between 1914 and 1968 skew towards the right. Only a small amount of data shows prices between 750 and 1000 USD. 

#The fourth subplot also doesn't make much sense, since it is a histogram of dates, but it was my only choice. obviously there is no skew and the median is right in the middle!
#I learned that a graph of years over the multiple years would be symmetrical- years pass at a constant rate. 
```
